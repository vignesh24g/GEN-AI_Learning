Gensim is an open-source Python library used for unsupervised topic modeling and natural language processing (NLP). It is designed to extract semantic topics from large volumes of text, convert text documents into numerical vectors, and find similarities between them.
Unlike general-purpose NLP libraries like spaCy or NLTK, Gensim is highly specialized for vector spaces, semantic computing, and word embeddings.
Key Capabilities of Gensim
• Efficient Word Embeddings: It provides fast, native implementations of popular vector models like Word2Vec, FastText, and Doc2Vec.

 Distance (Cosine Distance)model.distance(w1, w2) is simply the complement of cosine similarity:$$\text{distance}(\vec{u}, \vec{v}) = 1 - \text{similarity}(\vec{u}, \vec{v})$$What it means: It measures how far apart the two word vectors are oriented from each other.Range: Between $0$ and $2$ (typically between $0$ and $1$).Interpretation:Lower distance means closer meaning.Higher distance means unrelated concepts.

```python
model.distance('cricket','bat')
```

```python
model.distance('cricket','pencil')
```

model.similarity(w1, w2) computes the Cosine Similarity between the vectors $\vec{u}$ and $\vec{v}$ of the two words:

What it means: It evaluates whether the vectors point in roughly the same direction, regardless of vector magnitude.Range: Between $-1$ and $1$ (for most real-world word pairs, typically between $0$ and $1$).Interpretation:A value closer to 1.0 indicates strong contextual overlap (e.g., 'cricket' and 'bat' score 0.374).A value close to 0.0 indicates little to no contextual relationship (e.g., 'cricket' and 'pencil' score 0.058).

```python
model.similarity('cricket','bat')
```

```python
model.similarity('cricket','pencil')
```

The n_similarity function measures the semantic closeness between two groups of words (like two short phrases, sentences, or lists of keywords).
While model.similarity("cat", "dog") only compares one word to another, n_similarity allows you to compare entire blocks of text against each other.

List 1: ["sushi", "tuna"]   -->   Vector(sushi) + Vector(tuna)   -->   [Average Vector 1]
                                                                              │
                                                                       (Cosine Similarity)
                                                                              │
List 2: ["salmon", "rice"]  -->   Vector(salmon) + Vector(rice)  -->   [Average Vector 2]


```python
# Example 1: High semantic similarity (different words, same concept)
phrase_a = ["world", "cup", "soccer"]
phrase_b = ["football", "tournament", "match"]
score1 = model.n_similarity(phrase_a, phrase_b)
score1 # Expected: High score

```

```python
# Example 2: Low semantic similarity (completely different concepts)
phrase_c = ["corporate", "finance", "revenue"]
score2 = model.n_similarity(phrase_a, phrase_c)
score2  # Expected: Low score
```

Use code with caution.
Critical Rules & Limitations
• Order Doesn't Matter: Because it simply averages the vectors, ["dog", "bites", "man"] and ["man", "bites", "dog"] will yield the exact same score. It does not understand grammar or word order.
• Stopwords Matter: Common words like "the", "is", or "and" have their own vectors. If you leave them in your lists, they will dilute the average vector and skew your results. Always remove stopwords before using n_similarity.
• Out-of-Vocabulary (OOV) Errors: Every single word in both lists must exist in the Word2Vec vocabulary. If even one word is missing, the function will throw a KeyError.
